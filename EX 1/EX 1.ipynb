```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("driver_standings.csv")
```

```python
print("\nLast 5 Rows:")
print(df.tail())
```

```python
print("\nShape of Dataset:")
print(df.shape)
```

```python
print("\nColumn Names:")
print(df.columns)
```

```python
print("\nDataset Info:")
print(df.info())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
points = df['points'].values

print("\nArray:")
print(points)
```

```python
print("Sum =", np.sum(points))
print("Mean =", np.mean(points))
print("Maximum =", np.max(points))
print("Minimum =", np.min(points))
print("Standard Deviation =", np.std(points))
```

```python
points = df['points'].values

print("\nArray:")
print(points)
print("Sum =", np.sum(points))
print("Mean =", np.mean(points))
print("Maximum =", np.max(points))
print("Minimum =", np.min(points))
print("Standard Deviation =", np.std(points))
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['points'], bins=20, edgecolor='black')
plt.title("Histogram of Driver Points")
plt.xlabel("Points")
plt.ylabel("Frequency")
plt.show()
```

```python
top10 = df.nlargest(10, 'points')

plt.figure(figsize=(10,5))
plt.bar(range(len(top10)), top10['points'])
plt.xticks(range(len(top10)), top10['driverId'], rotation=45)
plt.title("Top 10 Driver Points")
plt.xlabel("Driver ID")
plt.ylabel("Points")
plt.show()
```
